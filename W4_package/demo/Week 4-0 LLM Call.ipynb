```python
import os
from openai import OpenAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
def call_llm(query):
    # 1. Initialize the OpenAI client
    # The client automatically picks up the OPENAI_API_KEY environment variable.
    client = OpenAI()

    # 2. Define the messages for the chat completion
    messages = [
        {"role": "system", "content": "You are a helpful and concise data science assistant."},
        {"role": "user", "content": query}
    ]

    print("Sending request to OpenAI API...\n")

    try:
        # 3. Make the Chat Completion call
        response = client.chat.completions.create(
            model="gpt-4o-mini",  # You can change this to "gpt-4o", etc.
            messages=messages,
            temperature=0.0
        )

        # 4. Analyze the Response
        print("--- ANALYSIS OF THE RESPONSE ---")
        
        # Extract basic message content
        choice = response.choices[0]
        message_content = choice.message.content
        finish_reason = choice.finish_reason
        
        print(f"Model Used: {response.model}")
        print(f"Finish Reason: {finish_reason} (e.g., 'stop' means complete, 'length' means hit max_tokens)")
        
        # Analyze token usage
        usage = response.usage
        if usage:
            print(f"\nToken Usage:")
            print(f"  - Prompt Tokens:     {usage.prompt_tokens}")
            print(f"  - Completion Tokens: {usage.completion_tokens}")
            print(f"  - Total Tokens:      {usage.total_tokens}")

        print("\n--- GENERATED CONTENT ---")
        print(message_content.strip())

    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
call_llm("What is a transformer model")
```
